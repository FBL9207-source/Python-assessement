```python
# # 1. Bank Account Manager

# a. Simulate a bank account with deposit, withdrawal, and balance enquiry operations
# b. Use variables and data types to store account information
# c. Implement conditional statements to check for sufficient balance before withdrawal
# d. Use functions to perform deposit, withdrawal, and balance inquiry operations.

balance = 0
def check_balance(balance):
    print("Current Balance: ",balance)
    return balance

def deposit(balance, amount):
    if amount <= 0:
        return balance
    balance += amount
    print(amount," deposited successfully.")
    print("New Balance: ",balance)
    return balance

def withdraw(balance, amount):
    if amount <= 0:
        print("Withdrawal amount must be positive.")
        return balance
    elif amount > balance:
        print("Insufficient balance")
        return balance
    else:
        balance -= amount
        print(amount," withdrawn successfully.")
        print("New Balance: ",balance)
        return balance
while True:
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        balance = check_balance(balance)
    elif choice == "2":
        amt = float(input("Enter amount to deposit: "))
        balance = deposit(balance, amt)
    elif choice == "3":
        amt = float(input("Enter amount to withdraw: "))
        balance = withdraw(balance, amt)
    elif choice == "4":
        print("Thank you for using Bank Account Manager. Goodbye!")
        break
    else:
        print("Invalid choice! Please select 1-4.")
```

```python
1# # 2. Weather Data Analyzer
# a. Analyze weather data (temperature, humidity, etc.) for a given region
# b. Use variables and data types to store weather data
# c. Implement conditional statements to determine weather conditions (sunny, rainy, etc.)
# d. Use functions to calculate average temperature and humidity, and print results

temperatures = [31, 29, 33, 27, 30, 34, 28]
humidity = [70, 65, 80, 60, 75, 55, 68]

def average_temperature(temp_list):
    avg = sum(temp_list) / len(temp_list)
    return avg

def average_humidity(hum_list):
    avg = sum(hum_list) / len(hum_list)
    return avg

def weather_condition(temp, hum):
    if temp >= 33 and hum >= 70:
        return "Hot and Humid"
    elif temp >= 33:
        return "Sunny"
    elif hum >= 70:
        return "Rainy"
    elif temp <= 28:
        return "Cool"
    else:
        return "Pleasant"

def analyze_weather(temp_list, hum_list):
    """Print full weather analysis"""
    avg_temp = average_temperature(temp_list)
    avg_hum = average_humidity(hum_list)

    print(f"Average Temperature: {avg_temp:.2f}°C")
    print(f"Average Humidity: {avg_hum:.2f}%")

    print("\nDay-wise Conditions:")
    for i in range(len(temp_list)):
        condition = weather_condition(temp_list[i], hum_list[i])
        print(f"Day {i+1}: Temp={temp_list[i]}°C, Humidity={hum_list[i]}% -> {condition}")
```

```python
# 3. Password Validator Write a function that checks if a password is strong: at least 8 characters, contains at least one digit, one uppercase letter, and one special character (!@#$%).

password = input("Enter a strong password")

upper = False
lower = False
digit = False
specail = False
spl_char = "!@#$%"
for letter in password:
  if letter.isupper():
    upper = True
  elif letter.islower():
    lower = True
  elif letter.isdigit():
    digit = True
  elif letter in spl_char:
    specail = True
if len(password) >= 8 and upper and lower and digit and specail:
  print("Strong password")
else:
  print("Weak password")
```

```python
# 4. Word Frequency Counter Given a paragraph of text, write a function that returns the 3 most frequently occurring words (ignore case and punctuation).


paragraph = input("enter a paragraph ")
paragraph = paragraph.lower()
punc = ",:'?."
word_frequency = {}
clean_paragraph = ""
for let in paragraph:
    if let not in punc:
        clean_paragraph += let
words = clean_paragraph.split()
for w in words:
    if w in word_frequency:
        word_frequency[w] += 1
    else:
        word_frequency[w] = 1
sorted_words = sorted(word_frequency.items(), key=lambda x: x[1], reverse=True)
top_3 = sorted_words[:3]
print("\n",top_3)
```

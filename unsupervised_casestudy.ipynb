```python
Q1_age = adult_df['age'].quantile(0.25)
Q3_age = adult_df['age'].quantile(0.75)
IQR_age = Q3_age - Q1_age

lower_bound_age = Q1_age - 1.5 * IQR_age
upper_bound_age = Q3_age + 1.5 * IQR_age

Q1_min_hours_week = adult_df['hours-per-week'].quantile(0.25)
Q3_min_hours_week = adult_df['hours-per-week'].quantile(0.75)
IQR_min_hours_week = Q3_min_hours_week - Q1_min_hours_week

lower_bound_min_hours_week = Q1_min_hours_week - 1.5 * IQR_min_hours_week
upper_bound_min_hours_week = Q3_min_hours_week + 1.5 * IQR_min_hours_week

Q1_min_education_num = adult_df['education-num'].quantile(0.25)
Q3_min_education_num = adult_df['education-num'].quantile(0.75)
IQR_min_education_num = Q3_min_education_num - Q1_min_education_num

lower_bound_min_education_num = Q1_min_education_num - 1.5 * IQR_min_education_num
upper_bound_min_education_num = Q3_min_education_num + 1.5 * IQR_min_education_num

Q1_min_fnlwgt = adult_df['fnlwgt'].quantile(0.25)
Q3_min_fnlwgt = adult_df['fnlwgt'].quantile(0.75)
IQR_min_fnlwgt = Q3_min_fnlwgt - Q1_min_fnlwgt

lower_bound_min_fnlwgt = Q1_min_fnlwgt - 1.5 * IQR_min_fnlwgt
upper_bound_min_fnlwgt = Q3_min_fnlwgt + 1.5 * IQR_min_fnlwgt

adult_df = adult_df[
    (adult_df['age'] > lower_bound_age) & (adult_df['age'] < upper_bound_age) &
    (adult_df['hours-per-week'] > lower_bound_min_hours_week) & (adult_df['hours-per-week'] < upper_bound_min_hours_week) &
    (adult_df['education-num'] > lower_bound_min_education_num) & (adult_df['education-num'] < upper_bound_min_education_num) &
    (adult_df['fnlwgt'] > lower_bound_min_fnlwgt) & (adult_df['fnlwgt'] < upper_bound_min_fnlwgt)
]
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
adult_df.select_dtypes(include=['int','float']).columns
```

```python
adult_df.select_dtypes(include=['object']).columns
```

```python
#Define columns

numeric_col = ['age', 'education-num', 'capital-gain', 'capital-loss',
       'hours-per-week']
categorical_col = ['workclass', 'marital-status', 'occupation',
       'relationship', 'race', 'sex', 'native-country']
```

```python
#Numerical pipeline
numerical_pipeline = Pipeline([('scaler',StandardScaler())])
```

```python
onehot_col = ['workclass', 'marital-status', 'occupation','relationship', 'race', 'sex', 'native-country']
```

```python
#categorical pipeline
categorical_pipeline = Pipeline([('encoder', OneHotEncoder(handle_unknown='ignore'))])
```

```python
#combine pipelines
preprocessor = ColumnTransformer([
    ('numeric',numerical_pipeline,numeric_col),
    ('onehot',categorical_pipeline,onehot_col)

])
```

```python
X = adult_df[numeric_col + onehot_col].copy()
income_labels = adult_df['income'].copy()
```

```python
#Fit and trnsform
X_processed = preprocessor.fit_transform(X)
```

```python
X_processed
```

```python
# using elbow method to find the optimum number of clusters
from sklearn.cluster import KMeans

wcss = []
for i in  range(3,9):
  kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42,n_init=10)
  kmeans.fit(X_processed)
  wcss.append (kmeans.inertia_)
```

```python
plt.plot(range(3,9),wcss,'o-')
plt.xlabel('Number of cluster')
plt.ylabel('wcss')
plt.grid()
```

```python
#elbow is appearing at k= 6
# we can create kmeans model with no of  clusters = 6

kmeans1 = KMeans(n_clusters=6, init='k-means++', random_state=42, n_init=10)
y_kmeans = kmeans1.fit_predict(X_processed)
```

```python
y_kmeans
```

```python
X_dense = X_processed.toarray()

plt.figure(figsize=(10, 6))

plt.scatter(X_dense[y_kmeans == 0, 0], X_dense[y_kmeans == 0, 1], c='red', label='Cluster 1')
plt.scatter(X_dense[y_kmeans == 1, 0], X_dense[y_kmeans == 1, 1], c='blue', label='Cluster 2')
plt.scatter(X_dense[y_kmeans == 2, 0], X_dense[y_kmeans == 2, 1], c='yellow', label='Cluster 3')
plt.scatter(X_dense[y_kmeans == 3, 0], X_dense[y_kmeans == 3, 1], c='green', label='Cluster 4')
plt.scatter(X_dense[y_kmeans == 4, 0], X_dense[y_kmeans == 4, 1], c='cyan', label='Cluster 5')
plt.scatter(X_dense[y_kmeans == 5, 0], X_dense[y_kmeans == 5, 1], c='black', label='Cluster 6')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering')
plt.legend()
plt.show()
```

```python
# stilhoutte score for kmeans clustering
from sklearn.metrics import silhouette_score
sil_kmeans = silhouette_score(X_processed,y_kmeans)
```
